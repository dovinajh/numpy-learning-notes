```python
import numpy as np
```

```python
#创建数组
arr=np.arange(1,13).reshape(3,4)
print(arr)
```

```python
#数组与数字作比较
print(arr>4)
```

```python
#筛选出arr>4的元素
print(arr[arr>4])
```

```python
import numpy as np
```

```python
#创建同维数组
arr1=np.arange(1,10)
arr2=np.flipud(arr1)
print(arr1)
print(arr2)
```

```python
#同维度数组作比较
print(arr1>arr2)
```

```python
#筛选出arr1>arr2位置上的元素
print(arr1[arr1>arr2])
print(arr2[arr1>arr2])
```
