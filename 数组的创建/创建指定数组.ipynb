```python
import numpy as np
```

```python
#创建一维数组--向量
arr1=np.array([1,2,3])
print(arr1)
```

```python
#创建二维数组--行矩阵
arr2=np.array([[1,2,3]])
print(arr2)
```

```python
#创建二维数组--列矩阵
arr3=np.array([[1],[2],[3]])
print(arr3)
```

```python
#创建二维数组--矩阵
arr4=np.array([[1,2,3],[4,5,6]])
print(arr4)
```
