```python
import numpy as np
```

```python
#创建矩阵
arr=np.arange(1,9).reshape(2,4)
print(arr)
```

```python
#按第一个维度（行）分裂
arr1,arr2=np.split(arr,[1])    #默认参数axis=0
print(arr1,'\n\n',arr2)        #注意输出的是矩阵
```

```python
#按第二个维度（列）分裂
arr1,arr2,arr3=np.split(arr,[1,3],axis=1)
print(arr1,'\n\n',arr2,'\n\n',arr3)
```
