```python
import numpy as np
```

```python
#创建数组1
arr1=np.array([[1,2,3],[4,5,6]])
print(arr1)
```

```python
#创建数组2
arr2=np.array([[7,8,9],[10,11,12]])
print(arr2)
```

```python
#按第一个维度（行）拼接
arr3=np.concatenate([arr1,arr2])    #默认参数axis=0
print(arr3)
```

```python
#按第二个维度（列）拼接
arr4=np.concatenate([arr1,arr2],axis=1)
print(arr4)
```
