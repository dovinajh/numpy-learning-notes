```python
import numpy as np
```

```python
#整数型数组
arr=np.array([1,2,3])
print(arr)
```

```python
#整数型数组与浮点数做运算
print(arr+0.0)
print(arr*1.0)
```

```python
#整数型数组遇到除法（即便是除以整数）
print(arr/1)
```

```python
#整数型数组与浮点数型数组做运算
int_arr=np.array([1,2,3])
float_arr=np.array([1.0,2,3])
print(int_arr+float_arr)
```
