```python
import numpy as np
```

```python
arr1=np.ones(3)    #传入形状3
print(arr1)       #造出一维数组
```

```python
arr2=np.ones((1,3))    #传入形状（1，3）
print(arr2)           #造出二维数组
```

```python
arr3=np.ones((1,1,3))    #传入形状（1，1，3）
print(arr3)             #造出三维数组
```
